```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
import joblib
import pandas as pd
```

```python
modelo_ocup = 'modelo_ocupacao_best_randomforestclassifier.pkl'
```

```python
modelo_carregado = joblib.load(modelo_ocup)

rf_model = modelo_carregado.named_steps["clf"]

# Pega a primeira árvore da floresta
tree = rf_model.estimators_[0]

# Nomes das features
feature_columns = ['Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio']

# Classes como strings
class_names = ["Vazio", "Ocupado"]

# Plot
plt.figure(figsize=(26, 30))
plot_tree(
    tree,
    feature_names=feature_columns,
    class_names=class_names,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.show()
```

```python
# Exemplo de medição
medicao = {
    "Temperature": [24.0],
    "Humidity": [40.0],
    "Light": [394.1],
    "CO2": [665.7],
    "HumidityRatio": [0.006303]
}

# Criar um DataFrame a partir dos dados do novo cliente
medicao_df = pd.DataFrame(medicao)

# Fazer a predição
predicao_enc = modelo_carregado.predict(medicao_df)
proba = modelo_carregado.predict_proba(medicao_df)

print(f"Predição: {predicao_enc}, prob.: {proba}")
```
